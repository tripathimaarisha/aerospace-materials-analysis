# Aerospace Materials: Python Visualisation

Exploratory Python visualisations of aerospace-alloy density/strength comparisons, conceptual low-temperature toughness profiles, and a synthetic precipitation-hardening curve.

## Scope and limitations

This is an exploratory computational-visualisation project. Some curves and parameters are synthetic or illustrative, not experimental measurements or validated predictions. Check all data sources, equations, units, and assumptions before making quantitative scientific claims.

## Python implementation

Run the following cell in Google Colab or Jupyter.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["axes.edgecolor"] = "#CCCCCC"
plt.rcParams["axes.linewidth"] = 0.8
plt.rcParams["grid.color"] = "#EAEAEA"
plt.rcParams["grid.linestyle"] = "--"

# Figure 1: Illustrative density-strength comparison.
# Verify the values against alloy datasheets/literature before quantitative use.
materials = {
    "Al 2024-T6\n(Apollo-era example)": [2.78, 345, "#1f77b4"],
    "Al 2024/2124-T74\n(Shuttle-era example)": [2.78, 310, "#ff7f0e"],
    "Al-Li 2195\n(lightweight tank alloy)": [2.70, 545, "#2ca02c"],
    "304L stainless steel\n(cryogenic structure example)": [7.93, 290, "#d62728"],
}
fig, ax = plt.subplots(figsize=(8, 6), dpi=120)
for label, (density, strength, color) in materials.items():
    ax.scatter(density, strength, s=130, color=color, edgecolors="black", linewidths=.5)
    ax.annotate(label, (density, strength), xytext=(8, 8),
                textcoords="offset points", fontsize=8)
ax.set(title="Aerospace alloy comparison: density and yield strength",
       xlabel=r"Density (g/cm$^3$)", ylabel="Yield strength (MPa)")
ax.grid(True)
plt.tight_layout()
plt.show()

# Figure 2: Synthetic conceptual curves, not measured toughness data.
temperature = np.linspace(-200, 20, 100)
carbon_steel = 120 / (1 + np.exp(-0.15 * (temperature + 40))) + 15
al_li_2195 = 110 - 0.12 * temperature
steel_304l = 140 - 0.05 * temperature
fig, ax = plt.subplots(figsize=(8, 5), dpi=120)
ax.plot(temperature, carbon_steel, "--", label="Illustrative carbon-steel curve")
ax.plot(temperature, al_li_2195, label="Illustrative Al-Li 2195 curve")
ax.plot(temperature, steel_304l, label="Illustrative 304L curve")
ax.axvspan(-196, -162, alpha=.25, label="Approximate liquid O2/methane range")
ax.set(title="Conceptual material-toughness profiles at low temperature",
       xlabel="Temperature (°C)", ylabel="Illustrative toughness (arbitrary units)")
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()

# Figure 3: Synthetic ageing curve for conceptual illustration only.
time = np.linspace(0, 100, 500)
strength_curve = 40 + 60 * (
    np.sin(time / 15) * np.exp(-time / 60)
    + 0.3 * (1 - np.exp(-time / 10))
)
strength_curve[250:] = strength_curve[250:] * .95 + 3.2
fig, ax = plt.subplots(figsize=(8, 5), dpi=120)
ax.plot(time, strength_curve, linewidth=2.5)
ax.set(title="Conceptual precipitation-hardening response",
       xlabel="Artificial-ageing time (arbitrary scale)",
       ylabel="Relative strength (arbitrary scale)")
ax.grid(True)
plt.tight_layout()
plt.show()
```
